```python

from glob import glob  # glob use to extract path of file

import cv2  # opencv
```

1. **Get list of all images path**

```python
fpath = glob('./data/female/*.jpg') # this will return all images path in a list
mpath = glob('./data/male/*.jpg')
```

```python
print('The number of images in Female folder = ',len(fpath))
print('The number of images in Male folder = ',len(mpath))
```

2. **Understand the 4 steps**

```python
### Step -1 Read Image and Convert to RGB
import matplotlib.pyplot as plt # visualze and display
img = cv2.imread(fpath[0])
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB ) # read image in BGR

# cv2.imshow('OpenCV', img_rgb)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

### Step -2: Apply Haar Cascade Classifier
haar = cv2.CascadeClassifier('./model/haarcascade_frontalface_default.xml')
gray = cv2.cvtColor(img_rgb, cv2.COLOR_BGR2GRAY)
face_rect = haar.detectMultiScale(gray, 1.5,5)
print(face_rect)

for x,y,w,h in face_rect:
    cv2.rectangle(img_rgb,(x,y),(x+w,y+h),(0,255,0),2)
    ### Step -3: crop Face
    roi = img_rgb[y:y+h,x:x+w]

    # cv2.imshow('Cropped Face', roi)
    # cv2.waitKey(0)
    # cv2.destroyAllWindows()
    plt.figure(figsize=(15,8))

    plt.subplot(1,2,1)
    plt.imshow(img_rgb)
    # plt.axis('off')

    plt.subplot(1,2,2)
    plt.imshow(roi)
    # plt.axis('off')

    plt.show()
```

---
3. Implement above process to all images

**Crop all `female` Images**

```python
haar = cv2.CascadeClassifier('./model/haarcascade_frontalface_default.xml')
```

```python
for i in range(len(fpath)):
    try:
        ### Step -1 Read Image and Convert to RGB
        img = cv2.imread(fpath[i]) # read image in BGR
        img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB) # this step will convert image from BGR to RGB

        ### Step -2: Apply Haar Cascade Classifier
        gray = cv2.cvtColor(img_rgb,cv2.COLOR_BGR2GRAY)
        faces_list = haar.detectMultiScale(gray,1.5,5)
        for x,y,w,h in faces_list:
            ### Step -3: crop Face
            roi = img[y:y+h,x:x+w]
            cv2.imwrite(f'./crop_data/female/female_{i}.jpg', roi)
            print("Image successfully processed: ", fpath[i])
    except:
        print("Unable to process image: ", fpath[i])
```

**Crop all `male` Images**

```python
for i in range(len(mpath)):
    try:
        ### Step -1 Read Image and Convert to RGB
        img = cv2.imread(mpath[i])
        img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)

        ### Step -2: Apply Haar Cascade Classifier
        gray = cv2.cvtColor(img_rgb,cv2.COLOR_BGR2GRAY)
        faces_list = haar.detectMultiScale(gray,1.5,5)
        for x,y,w,h in faces_list:
            ### Step -3: crop Face
            roi = img[y:y+h,x:x+w]
            cv2.imwrite(f'./crop_data/male/male_{i}.jpg', roi)
            print('Image Sucessfully processed')
    except:
        print('Unable to Process the image')
```
